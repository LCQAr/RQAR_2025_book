## 1. Detectar as flags internas nos dados

```python
import os 
import pandas as pd
import warnings
import numpy as np
warnings.filterwarnings('ignore')

# Caminho para a pasta de dados
rootPath = os.path.dirname(os.getcwd())

# Lendo o csv
aqmData = pd.read_csv(rootPath+'/data/Monitoramento_QAr_BR.csv')

col = 'QAQC_INTERNO'
if col not in aqmData.columns:
    aqmData[col] = ""   # string vazia como preenchimento
    
# Verifica se o arquivo com os dados existe
possible_flags = []
tem_qaqc = []

for index, row in aqmData.iterrows():
    if pd.isna(row.ID_MMA_COMPLETO) | pd.isna(row.POLUENTE): 
        #print('ID_MMA não foi criado')
        tem_qaqc.append(False)
    else:
        file_path = rootPath+'/data/MQAr/'+row.POLUENTE+'/'+row.ID_MMA_COMPLETO + '.csv'
        if os.path.exists(file_path):
            #print(f"'{file_path}' existe.")
            df = pd.read_csv(file_path)
            
            # Verifica se a coluna valor existe
            if 'QAQC_INTERNO' in df.columns:
             #   print("Coluna QAQC_INTERNO existe")
                possible_flags.append(df['QAQC_INTERNO'].unique())
                if df['QAQC_INTERNO'].unique().any() != '' :
                    tem_qaqc.append(True)
                else:
                    tem_qaqc.append(False)
            else:
                print("Coluna QAQC_INTERNO NÃO existe")
                tem_qaqc.append(False)
                
        else:
            #print(f"'{file_path}' não existe.")
            tem_qaqc.append(False)

aqmData['QAQC_INTERNO'] = tem_qaqc
possible_flags = np.unique(possible_flags)
exploded_list = []
for sub_array in possible_flags:
    for item in sub_array:
        exploded_list.append(item)
        
possible_flags = np.unique(exploded_list)
```

## 2. Criar dicionário de tradução das possíveis flags

1. Brasil – CETESB / QUALAR

CETESB (Companhia Ambiental do Estado de São Paulo)
Manual do Sistema QUALAR (Qualidade do Ar).
👉 Esse sistema utiliza códigos de validação como Inválido, Insuficientes, Inexistente, Fora da Faixa de Medição, além de flags alfanuméricas (I*, V*) para rastrear a origem do problema.

Fonte: CETESB – QUALAR - Sistema de Qualidade do Ar (documentação interna e relatórios técnicos).

2. EMEP/EEA – European Environment Agency

EMEP/EEA Air Quality Data Flagging Codes
Usado em bancos como AirBase (agora integrado ao European Air Quality Portal).
👉 Define flags como < (abaixo do limite de detecção), > (acima do limite), L (low detection), N (não disponível), F (falha), P (provisório).

Fonte: EEA, Air Quality e-Reporting (AQ e-Reporting) Guidelines.

3. USEPA – United States Environmental Protection Agency

EPA Air Quality System (AQS) Data Coding Manual
👉 Usa códigos para indicar validade, detecção abaixo/above range, flags de QA/QC e dados substituídos.

Fonte: USEPA, Air Quality System (AQS) Data Coding Manual (EPA-454/R-13-005).

4. Outras fontes técnicas

WHO Air Quality Monitoring Guidelines → traz recomendações gerais de validação.

Norma ABNT NBR 17025 (ensaios e calibração) → sobre rastreabilidade e invalidação de dados.

📌 Em resumo:

As flags textuais em português vêm do QUALAR (CETESB).

As flags simbólicas (<, >, =, !) têm origem em EEA/USEPA.

As variações I* e V* são tipicamente CETESB/QUALAR, que detalha o motivo da invalidação/validação.


| Categoria                        | CETESB / QUALAR                                              | EEA / EMEP                      | USEPA / AQS                         | Significado                                                       |
| -------------------------------- | ------------------------------------------------------------ | ------------------------------- | ----------------------------------- | ----------------------------------------------------------------- |
| Valor válido                     | `" "` (vazio), `"A"`, `"VA"`, `"VC"`, `"VR"`, `"VS"`         | Sem flag                        | Valid                               | Dado considerado válido                                           |
| Provisório                       | `"P"`, `"VU"`                                                | `P`                             | Preliminary                         | Dado ainda não validado manualmente                               |
| Suspeito                         | `"!"`, `"?E"`, `"?S"`, `"?U"`                                | `S` (suspect)                   | Suspect flag                        | Dado suspeito, sob revisão                                        |
| Abaixo do limite de detecção     | `"L"`, `"<"`                                                 | `<`                             | `<`                                 | Valor medido abaixo do LOD                                        |
| Acima do limite de detecção      | `">"`, `"Fora da Faixa de Medição"`                          | `>`                             | `>`                                 | Valor acima da faixa de medição                                   |
| Falha / erro                     | `"F"`, `"I"`, `"Inválido"`, `"Disabilitada Temporariamente"` | `F` (fail)                      | `I` (invalid)                       | Dado inválido ou falha no equipamento                             |
| Motivo específico de invalidação | `"IC"`, `"IM"`, `"IR"`, `"IS"`, etc.                         | Não detalhado (apenas inválido) | `Qualifier Codes` (ex.: `QC`, `IN`) | Rastreia a origem da falha (calibração, sensor, manutenção, rede) |
| Não disponível                   | `"N"`, `"Inexistente"`                                       | `N`                             | `NA`                                | Dado não existe ou não foi registrado                             |
| Dados insuficientes              | `"Insuficientes"`                                            | Não existe                      | Incomplete                          | Não há dados suficientes para estatística                         |
| Corrigido / filtrado             | `"f"`, `"VE"`                                                | `C` (corrected)                 | Substituted Value                   | Valor corrigido ou estimado                                       |

```python
flags_validacao = {
    # Dado válido, sem observação
    " ": True,

    # Alerta, valor suspeito
    "!": False,

    # Abaixo do limite de detecção
    "<": True,

    # Valor repetido ou substituído
    "=": True,

    # Acima do limite de detecção
    ">": True,

    # Dúvida de erro de equipamento
    "?E": False,

    # Dúvida de erro de sistema/software
    "?S": False,

    # Incerteza desconhecida
    "?U": False,

    # Dado validado automaticamente
    "A": True,

    # Estação/parâmetro fora de operação
    "Disabilitada Temporariamente": False,

    # Falha de sensor
    "F": False,

    # Extrapolou o range do sensor
    "Fora da Faixa de Medição": False,

    # Inválido (genérico)
    "I": False,

    # Subtipos de inválido
    "I#": False,  # Inválido - erro não especificado
    "I%": False,  # Inválido - erro de cálculo/percentual
    "IC": False,  # Inválido - calibração
    "ID": False,  # Inválido - dado duplicado
    "IE": False,  # Inválido - erro de equipamento
    "IF": False,  # Inválido - falha de funcionamento
    "IG": False,  # Inválido - erro genérico
    "II": False,  # Inválido - indefinido
    "IL": False,  # Inválido - abaixo do limite
    "IM": False,  # Inválido - manutenção
    "IP": False,  # Inválido - processamento
    "IR": False,  # Inválido - rede/comunicação
    "IS": False,  # Inválido - sensor
    "IT": False,  # Inválido - tempo inconsistente
    "IU": False,  # Inválido - motivo desconhecido
    "IZ": False,  # Inválido - outros motivos

    # Parâmetro não medido na estação
    "Inexistente": False,

    # Poucos dados para estatística
    "Insuficientes": False,

    # Dado descartado após validação
    "Inválido": False,

    # Abaixo do limite de detecção
    "L": True,

    # Não disponível
    "N": False,

    # Dado provisório (sem validação manual)
    "P": True,

    # Flags de validação manual
    "VA": True,  # Validado automaticamente
    "VC": True,  # Validado por critério específico
    "VE": True,  # Validado - erro corrigido
    "VG": True,  # Validado - genérico
    "VI": True,  # Validado por inspeção
    "VR": True,  # Validado e revisado
    "VS": True,  # Validado pelo sistema
    "VU": True,  # Validado com incerteza

    # Dado filtrado/corrigido
    "f": True,

    # Valor ausente/não numérico
    "nan": False,

    # Dado inválido, não utilizar
    "x": False
}
```

## 3. Atualizar o nome das flags



```python
   
# Verifica se o arquivo com os dados existe
for index, row in aqmData.iterrows():
    if pd.isna(row.ID_MMA_COMPLETO) | pd.isna(row.POLUENTE): 
        print('ID_MMA não foi criado')
        
    else:
        file_path = rootPath+'/data/MQAr/'+row.POLUENTE+'/'+row.ID_MMA_COMPLETO + '.csv'
        if os.path.exists(file_path):
            df = pd.read_csv(file_path)  
            col = 'QAQC_MMA'
            if col not in aqmData.columns:
                df[col] = ""   # string vazia como preenchimento
            print(f"'{file_path}' existe.")
            
            df["QAQC_MMA"] = df["QAQC_INTERNO"].map(flags_validacao)
            df.to_csv(file_path,index=False)
        else:
            print(f"'{file_path}' não existe.")
            


```
