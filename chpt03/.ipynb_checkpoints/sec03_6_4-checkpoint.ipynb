### 3.6.4. Cobertura populacional

<div style="text-align: justify">A estimativa da cobertura populacional das estações de monitoramento foi realizada com base em dados populacionais distribuídos em setores censitários estabelecidos no Censo populacional de 2022 disponibilizados pelo IBGE <cite id="q2dvr"><a href="#zotero%7C22267313%2F43V9SKCC">(IBGE, 2022)</a></cite>, integrados às áreas de representatividade espacial (seção 3.6.1) de cada estação. O objetivo é estimar a população coberta por cada ponto da rede de monitoramento da qualidade do ar. 
    
</p>A população atendida foi estimada por meio da interseção espacial entre os buffers e os setores censitários. Para cada setor que se sobrepôs a um buffer, calculou-se a fração da área do setor contida no polígono e aplicou-se essa proporção à população total do setor.
